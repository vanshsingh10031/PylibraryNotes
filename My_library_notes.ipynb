```python
DataFrame.iterrows(), 
#which can loop a DataFrame row-wise. 
#It returns the index and row of the DataFrame in each iteration of the for a loop.
#examples
import pandas as pd

# Create DataFrame from dict
student_dict = {'Name': ['Joe', 'Nat'], 'Age': [20, 21], 'Marks': [85, 77]}
student_df = pd.DataFrame(student_dict)

# Iterate all the rows of DataFrame
for index, row in student_df.iterrows():
    print(index, row)
```
